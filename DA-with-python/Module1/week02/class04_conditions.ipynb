# Conditions

```python
# if / elif / else -- runs one block based on which condition is True

# syntax:
# if condition:
#     logic
# elif another_condition:
#     logic
# else:
#     logic
```

```python
# CNIC eligibility: age >= 18 -> eligible, else not eligible

age = int(input("Enter your age: "))
print(age)


if age >= 18:
    print("Yes! you are eligible for the cnic....")
    
else: # runs only when the if condition is False
    print("No! You are not eligble... wait ___ years")
```

```python
# Marks grading:
# 90+     -> A+
# 80-89   -> A
# 70-79   -> B
# 50-69   -> C
# below 50 -> Failed

# marks = int(input("Enter your marks..."))
# print(marks)

marks = 90

if marks >= 90: # grade A+
    print("You got grade A+ Congratulations!")

elif marks >= 80 and marks < 90: # grade A
    print("You got grade A+ Congratulations!") 

elif marks >= 70 and marks < 80: # grade B
    print("Grade B.. Keep it up!") 

elif marks >= 50 and marks < 70: # grade C
    print("grade C ... do hard work")

else: # anything below 50
    print("Failed ... retry")
```

### Operator - in

```python
numbers = [1,2,3,4,5,6,7] 

num = int(input("Enter a number... "))

# in -- checks whether num exists anywhere in the list
if num in numbers:
    print("yes")
else:
    print("no")
```
