Q2 — Adding & Updating Values

car = {
"brand": "Toyota",
"model": "Corolla",
"year": 2018
}

Add color='White', update year to 2022 and print dictionary.



```python
car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2018
}

# Add a new key and value
car["color"] = "White"

# Update year
car["year"] = 2022

print(car)
```

Q3 — Loop Through Dictionary

person = {
"name": "Ali",
"age": 25,
"city": "Karachi"
}

Print keys and values using loop

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

for key, value in person.items():
    print(key, ":", value)
```

Q4 — Count Frequency of Numbers

numbers = [1,2,2,3,3,3,4,4,4,4]

Create frequency dictionary

```python
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)
```

Q5 — Word Counter

Take sentence input and count word frequency.

```python
sentence = input("Enter a sentence: ")

words = sentence.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

Q6 — Student Marks & Average

marks = {"Math":80, "English":70, "Physics":90}

Calculate total and average.

```python
marks = {
    "Math": 80,
    "English": 70,
    "Physics": 90
}

total = sum(marks.values())
average = total / len(marks)

print("Total:", total)
print("Average:", average)
```

Q7 — Highest Scoring Student

students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}

Find highest scoring studen

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}

highest_student = max(students, key=students.get)

print("Highest scoring student:", highest_student)
print("Marks:", students[highest_student])
```
